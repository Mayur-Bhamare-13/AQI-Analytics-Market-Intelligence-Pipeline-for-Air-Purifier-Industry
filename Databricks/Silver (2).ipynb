```python
from pyspark.sql.functions import col, lit

if spark.catalog.tableExists("aircat.silver.aqi_silver"):
    max_aid = spark.sql("SELECT COALESCE(MAX(aid),0) FROM aircat.silver.aqi_silver").first()[0]
else:
    max_aid = 0

aqi_silver_df = spark.sql("""
    SELECT *,
    ROW_NUMBER() OVER (ORDER BY processed_timestamp) AS rn
    FROM aircat.bronze.aqi_bronze
""").withColumn("aid", col("rn") + lit(max_aid)).drop("rn")

aqi_silver_df=aqi_silver_df.withColumnRenamed('month','year_').withColumnRenamed('year','month')

aqi_silver_df.createOrReplaceTempView('aqi_silver_temp')
```

```python
%sql
create table if not exists aircat.silver.aqi_silver
(
   year_ int,
   month string,
   state string,
   area string,
   aqi_value double,
   number_of_monitoring_stations integer,
   processed_timestamp timestamp,
   aid integer
)
USING DELTA
```

```python
# Transforming health table, because it had multiple entries of same date, aggregating all the info into a same single date as follows:

from pyspark.sql.functions import col, lit

if spark.catalog.tableExists("aircat.silver.health_silver"):
    max_hid = spark.sql("SELECT COALESCE(MAX(hid),0) FROM aircat.silver.health_silver").first()[0]
else:
    max_hid = 0

spark.sql("""
    SELECT *,
    ROW_NUMBER() OVER (ORDER BY processed_timestamp) AS rn
    FROM 
    (
            select 
            year, 
            reporting_date, 
            state, 
            disease_illness_name,
            reporting_month,
            sum(cases) as cases,
            sum(deaths) as deaths,
            max(processed_timestamp) as processed_timestamp
            from aircat.bronze.health_bronze
            where reporting_date is not null
            group by 
            year, 
            reporting_date, 
            state, 
            disease_illness_name,
            reporting_month
    )
    
""").withColumn("hid", col("rn") + lit(max_hid)).drop("rn").createOrReplaceTempView('health_silver_temp')
```

```python
%sql
create table if not exists aircat.silver.health_silver
(
    year int,
    reporting_date date,
    state string,
    disease_illness_name string,
    cases int,
    deaths int,
    reporting_month string,
    processed_timestamp timestamp,
    hid int
)
USING DELTA
```

Merge/Upsert into aqi_silver

```python
%sql
MERGE INTO aircat.silver.aqi_silver AS target
USING (
    SELECT *
    FROM aqi_silver_temp
    WHERE processed_timestamp > (
        SELECT COALESCE(MAX(processed_timestamp), '1900-01-01')
        FROM aircat.silver.aqi_silver
    )
) AS source

ON target.state = source.state
AND target.area=source.area
AND target.year_ = source.year_
AND target.month = source.month

WHEN MATCHED 
AND source.processed_timestamp > target.processed_timestamp
THEN UPDATE SET 
  target.year_ = source.year_,
  target.month = source.month,
  target.state = source.state,
  target.area = source.area,
  target.aqi_value = source.aqi_value,
  target.number_of_monitoring_stations = source.number_of_monitoring_stations,
  target.processed_timestamp = source.processed_timestamp

WHEN NOT MATCHED THEN
  INSERT *
```
