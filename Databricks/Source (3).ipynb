creating aqi source

```python
spark.read.csv('/Volumes/aircat/default/airvolume/aqi.csv', inferSchema=True, header=True)\
  .drop('_c0')\
  .createOrReplaceTempView('aqi_source_temp')

```

```python
%sql
    
create table if not exists aircat.source.aqi_source(
  month int,
  year string,
  state string,
  area string,
  aqi_value double,
  number_of_monitoring_stations int,
  processed_timestamp timestamp
)
```

```python
%sql
insert into aircat.source.aqi_source
select *,
 current_timestamp() as processed_timestamp
from aqi_source_temp
```

```python
%sql
select * from aircat.source.aqi_source
```
